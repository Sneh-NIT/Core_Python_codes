```python
# 2nd question
ac_no_list = df['AC No'].unique()
election_year_list = df['Election Year'].unique()
swing_seat = 0
for ac_no in ac_no_list:
  party_list = []
  for election_year in election_year_list:
    if not df.loc[(df['AC No']==ac_no) & (df['Election Year'] == election_year)].empty:
      idx = df.loc[(df['AC No']==ac_no) & (df['Election Year'] == election_year)]['Votes'].idxmax()
      party_list.append(df.loc[idx]['Party'])
  print(len(party_list))
  if party_list:
    if len(party_list) == 4:
      if party_list[0] != party_list[1] and party_list[0] != party_list[2] and party_list[0] != party_list[3]:
        swing_seat += 1
print("Swing seats: ",swing_seat)
```

```python
#3rd question Determine the top 8 parties at the state-level in each election in terms of getting
# the most votes.
party_unique_list = df['Party'].unique()
election_year_list = df['Election Year'].unique()
party_election_year_votes_df = pd.DataFrame(columns=['Party','Election Year','Votes_sum'])
output_df = party_election_year_votes_df.copy()
for election_year in election_year_list:
  for party in party_unique_list:
    party_election_year_votes_sum = df.loc[(df['Election Year'] == election_year) & (df['Party']==party)]['Votes'].sum()
    new_row = pd.DataFrame({'Party':party,'Election Year':election_year,'Votes_sum':party_election_year_votes_sum},index=[0])
    party_election_year_votes_df = pd.concat([party_election_year_votes_df, new_row],ignore_index=True)

party_election_year_votes_df_sorted = party_election_year_votes_df.sort_values(by=['Election Year','Votes_sum'],ascending=False)
for election_year in election_year_list:
   for i in range(8):
    idx = party_election_year_votes_df_sorted.loc[party_election_year_votes_df_sorted['Election Year']==election_year]['Votes_sum'].idxmax()
    new_row = pd.DataFrame({'Party':party_election_year_votes_df_sorted.loc[idx,'Party'],'Election Year':party_election_year_votes_df_sorted.loc[idx,'Election Year'], 'Votes_sum':party_election_year_votes_df_sorted.loc[idx,'Votes_sum']}, index=[0])
    output_df = pd.concat([output_df,new_row],ignore_index=True)
    party_election_year_votes_df_sorted = party_election_year_votes_df_sorted.drop(idx)
output_df
```

```python
#4th question
# Determine the minimum votes that each party in the top 8 of the 2015 election
# secured among all the past elections. Do this for all constituencies.
party_list = output_df.loc[(output_df['Election Year']=='2015')]
output_4th_df = pd.DataFrame(columns=['Party','Election Year','Min_Votes'])
for party in party_list['Party']:
  idx = df.loc[(df['Party']==party) & (df['Election Year']=='2015')]['Votes'].idxmin()
  new_row = pd.DataFrame({'Party':df.loc[idx,'Party'],'Election Year':df.loc[idx,'Election Year'], 'Min_Votes':df.loc[idx,'Votes']}, index=[0])
  output_4th_df = pd.concat([output_4th_df,new_row],ignore_index=True)
output_4th_df
```

```python
# 5th question : Determine the consistently tri-polar constituencies. I.e Constituencies in which 3
# parties or more got greater than 10% of the vote shares in each election. The
# parties can be different for different elections.
ac_no_list = df['AC No'].unique()
election_year_list = df['Election Year'].unique()
party_unique_list = df['Party'].unique()
output_5th_df = pd.DataFrame(columns=['AC No','AC name','Election Year'])
for ac_no in ac_no_list:
    for election_year in election_year_list:
      party_count = 0
      for party in party_unique_list:
        if not df.loc[(df['AC No']==ac_no)&(df['Party']==party)&(df['Election Year']==election_year)]['Votes %'].empty:
          # print("AC No: ", ac_no, "Party: ", party, "Election year: ", election_year, "Total Votes % Sum: ", df.loc[(df['AC No']==ac_no)&(df['Party']==party)&(df['Election Year']==election_year)]['Votes %'].sum())
          if df.loc[(df['AC No']==ac_no)&(df['Party']==party)&(df['Election Year']==election_year)]['Votes %'].sum() > 10.00:
            party_count += 1
      if party_count >= 3:
        print('AC No: ', ac_no,'AC name: ', df.loc[df['AC No']==ac_no]['AC name'].unique()[0], 'Election Year: ', election_year)
        new_row = pd.DataFrame({'AC No':ac_no,'AC name':df.loc[df['AC No']==ac_no]['AC name'].unique()[0],'Election Year':election_year},index=[0])
        output_5th_df = pd.concat([output_5th_df,new_row],ignore_index=True)
output_5th_df
```

```python
# 8th question
# Find the growing seats. For the top 8 parties at the state-level in the 2005-Feb
# Election, Determine the seats in which these parties have continuously increased
# their votes in the next 3 elections.
feb_2005_df = df.loc[df['Election Year']=='2005-Feb']
ac_no_list = df['AC No'].unique()
party_list = feb_2005_df.groupby(by=['Party'])['Votes'].sum().sort_values(ascending=False)[:9].index.tolist()
oct_2005_df = df.loc[df['Election Year']=='2005-Oct']
df_2010 = df.loc[df['Election Year']=='2010']
df_2015 = df.loc[df['Election Year']=='2015']
party_ac_no_dict = {}
for party in party_list:
  party_ac_no_dict[party] = []
for ac_no in ac_no_list:
  for party in party_list:
    feb_2005_value = df.loc[(df['AC No']==ac_no)&(df['Election Year']=='2005-Feb')&((df['Party']==party))]['Votes'].sum()
    oct_2005_value = df.loc[(df['AC No']==ac_no)&(df['Election Year']=='2005-Oct')&((df['Party']==party))]['Votes'].sum()
    value_2010 = df.loc[(df['AC No']==ac_no)&(df['Election Year']=='2010')&((df['Party']==party))]['Votes'].sum()
    value_2015 = df.loc[(df['AC No']==ac_no)&(df['Election Year']=='2015')&((df['Party']==party))]['Votes'].sum()
    if value_2015>value_2010>oct_2005_value>feb_2005_value:
      party_ac_no_dict[party].append(int(ac_no))
print(party_ac_no_dict)
```

```python
# Question 9: Find the declining seats. Repeat exercise g for the opposite growth.
ac_no_list = df['AC No'].unique()
election_year_list = df['Election Year'].unique()
output_7th_df = pd.DataFrame(columns=['AC No','AC name'])
for ac_no in ac_no_list:
  bottom_two_parties = []
  for election_year in election_year_list:
    bottom_two_parties_election_year = []
    if not df.loc[(df['AC No']==ac_no)&(df['Election Year']==election_year)]['Votes'].empty:
      idx1,idx2 = df.loc[(df['AC No']==ac_no)&(df['Election Year']==election_year)]['Votes'].nsmallest(2).index
      votes_1 = False
      votes_2 = False
      if df.loc[idx1]['Votes %']:
        votes_1 = True
      if df.loc[idx2]['Votes %']:
        votes_2 = True
      if votes_1 == True and votes_2 == True and Decimal(df.loc[idx1]['Votes %'])-Decimal(df.loc[idx2]['Votes %']) < 10.00:
        bottom_two_parties_election_year.append((df.loc[idx1]['Party'], df.loc[idx2]['Party']))
        bottom_two_parties.append(bottom_two_parties_election_year)
  print(bottom_two_parties)
```

```python
# Question 10: Determine the state-winner-match constituencies. Find the seats which have
# always elected the party that has gone on to win the state election (most votes
# secured at state). So if the RJD won at the state level in 2010, then seats which
# voted an RJD candidate as winner are considered in this category. This criteria
# for a seat has to be matched for each election.
feb_2005_df = df.loc[df['Election Year']=='2005-Feb']
print("Feb 2005 elections winner: ",feb_2005_df.groupby(by=['Party'])['Votes'].sum().idxmax()," votes: ", feb_2005_df.groupby(by=['Party'])['Votes'].sum().max())
oct_2005_df = df.loc[df['Election Year']=='2005-Oct']
print("Oct 2005 elections winner: ",oct_2005_df.groupby(by=['Party'])['Votes'].sum().idxmax()," votes: ", oct_2005_df.groupby(by=['Party'])['Votes'].sum().max())
df_2010 = df.loc[df['Election Year']=='2010']
print("2010 elections winner: ",df_2010.groupby(by=['Party'])['Votes'].sum().idxmax()," votes: ", df_2010.groupby(by=['Party'])['Votes'].sum().max())
# print(df_2010.groupby(by=['Party'])['Votes'].sum())
df_2015 = df.loc[df['Election Year']=='2015']
print("2015 elections winner: ",df_2015.groupby(by=['Party'])['Votes'].sum().idxmax()," votes: ", df_2015.groupby(by=['Party'])['Votes'].sum().max())

# Feb 2005 elections winner:  Rashtriya Janata Dal  votes:  6140223
# Oct 2005 elections winner:  Rashtriya Janata Dal  votes:  5525081
# 2010 elections winner:  Janata Dal (United)  votes:  6561906
# 2015 elections winner:  Bharatiya Janta Party  votes:  9308015
ac_no_list = df['AC No'].unique()
state_winner_match = {}
state_winner_match['Rashtriya Janata Dal-Feb 2005']=[]
state_winner_match['Rashtriya Janata Dal-Oct 2005']=[]
state_winner_match['Janata Dal (United)-2010']=[]
state_winner_match['Bharatiya Janta Party-2015']=[]
for ac_no in ac_no_list:
  if not df.loc[(df['Election Year']=='2005-Feb')&(df['AC No']==ac_no)]['Votes'].empty:
    idx = df.loc[(df['Election Year']=='2005-Feb')&(df['AC No']==ac_no)]['Votes'].idxmax()
    winner_row = df.loc[idx]
    if winner_row['Party']=='Rashtriya Janata Dal':
      state_winner_match['Rashtriya Janata Dal-Feb 2005'].append(int(winner_row['AC No']))
  if not df.loc[(df['Election Year']=='2005-Oct')&(df['AC No']==ac_no)]['Votes'].empty:
    idx = df.loc[(df['Election Year']=='2005-Oct')&(df['AC No']==ac_no)]['Votes'].idxmax()
    winner_row = df.loc[idx]
    if winner_row['Party']=='Rashtriya Janata Dal':
      state_winner_match['Rashtriya Janata Dal-Oct 2005'].append(int(winner_row['AC No']))
  if not df.loc[(df['Election Year']=='2010')&(df['AC No']==ac_no)]['Votes'].empty:
    idx = df.loc[(df['Election Year']=='2010')&(df['AC No']==ac_no)]['Votes'].idxmax()
    winner_row = df.loc[idx]
    if winner_row['Party']=='Janata Dal (United)':
      state_winner_match['Janata Dal (United)-2010'].append(int(winner_row['AC No']))
  if not df.loc[(df['Election Year']=='2015')&(df['AC No']==ac_no)]['Votes'].empty:
    idx = df.loc[(df['Election Year']=='2015')&(df['AC No']==ac_no)]['Votes'].idxmax()
    winner_row = df.loc[idx]
    if winner_row['Party']=='Bharatiya Janta Party':
      state_winner_match['Bharatiya Janta Party-2015'].append(int(winner_row['AC No']))
print(state_winner_match)
```
