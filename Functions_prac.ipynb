```python
def lesser_of_two_evens(a, b):
    if a <= b:
        return a
    else:
        return b
```

```python
lesser_of_two_evens(4,5)
```

```python
def makes_twenty(a, b):
    if a + b == 20 or a == 20 or b == 20:
        return True
    else:
        return False
```

```python
makes_twenty(50, 30)
```

```python
def old_macdonald(str1):
    str1 = str1[0].upper() + str1[1:3] + str1[3].upper() + str1[4:]
    return str1
```

```python
old_macdonald('macdonald')
```

```python
def master_yoda(str1):
    return " ".join(str1.split(" ")[::-1])
```

```python
master_yoda("I am home")
```

```python
def almost_there(n):
    if (n - 100 >= -10 and n - 100 <= 0) or (n - 100 >=0 and n - 100 <= 10) or (n - 200 >= -10 and n - 100 <= 0) or (n - 200 >= 0 and n - 200 <= 10): 
        return True
    else:
        return False
```

```python
def has_33(list1):
    count = 0
    for ele in list1:
        if ele == 3:
            count += 1
        else:
            count = 0
        if count == 2:
            return True
    return False
```

```python
has_33([1,3,3])
```

```python
def paper_doll(text):
    new_str = ""
    for ele in text:
        new_str = new_str + ele * 3
    return new_str
```

```python
text = "Hello"
```

```python
paper_doll(text)
```

```python
def blackjack(a, b, c):
    sum_val = a + b + c
    if sum_val <= 21:
        return sum_val
    else:
        sum_val = sum_val - 10
        if (a == 11 or b == 11 or c == 11):
            return sum_val
        if sum_val > 21:
                print("BUST")
```

```python
blackjack(10, 23,21)
```

```python
def summer_69(list1):
    if not list1:
        print(list1)
        return 0
    else:
        sum_val = 0
        for ele in list1:
            if ele in (6,7,8,9):
                continue
            sum_val += ele
        return sum_val
```

```python
summer_69([2,1,6,9,11])
```

```python
def spy_game(nums):
    count_seq = 0
    for ele in nums:
        if count_seq == 2:
            if ele == 7:
                return True
        if ele == 0:
            count_seq += 1
        if ele == 7:
            count_seq -=1
    return False
```

```python
spy_game([1,7,2,0,4,5,0])
```

```python
def count_primes(num):
    count_prime = 0
    for ele in range(2, num+1):
        prime_flag = True
        for sub_ele in range(2, ele):
            if ele % sub_ele == 0:
                prime_flag = False
                break
        if prime_flag == True:
            count_prime += 1
    return count_prime
```

```python
count_primes(100)
```
