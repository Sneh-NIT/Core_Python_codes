```python
name = "programming"
count_vowel = 0
count_consonant = 0
for ele in name:
    if ele == 'a' or ele == 'e' or ele == 'i' or ele == 'o' or ele == 'u':
        count_vowel += 1
    else:
        count_consonant += 1
print(count_vowel)
print(count_consonant)
```

```python
x = 0
count_vowel = 0
count_consonant = 0
while x in range(len(name)):
    if name[x] == 'a' or name[x] == 'e' or name[x] == 'i' or name[x] == 'o' or name[x] == 'u':
        count_vowel += 1
    else:
        count_consonant += 1
    x += 1
print(count_vowel)
print(count_consonant)
```

```python
# Write a programm which checks a number is palindrome
# Write a while loop that finds sum of 1 to n numbers, n input by user.
#Code using while loop
original_num = 873328
n = original_num
ele = 1
new_num = 0
while n != 0:
    ele = n % 10
    new_num = 10*new_num + ele
    n = n // 10
if new_num == original_num:
    print(f"The {new_num} is a palindrome")
else:
    print(f"The {new_num} is not a palindrome")
```
